```python
import matplotlib.pyplot as plt

day1=[1,2,3,4,5,6]
grwothLevel=[20,40,47,78,23,56]

plt.bar(day1,grwothLevel,label="Growth")

plt.title("Day Vs Growth",fontsize=24)
plt.xlabel("<------------ Days -------->",fontsize=24)
plt.ylabel("<------------ Growth -------->",fontsize=24)
plt.legend()
plt.show()
```
